**Abstract**:
This notebook demonstrates how Jupyter notebooks can be used to compute data provenance for a sequence of SQL queries. To achieve this, each SQL query is first translated into a set of (functional) tuple-generating dependencies, providing a formal logical representation of the query semantics. This intermediate representation serves as the foundation for a variety of logical and structural optimizations. Finally, the optimized f-tgds are back-parsed into nested SQL queries, which are then used to compute the provenance of the resulting query outputs.

### SQL -> f-tgd
Parsing a SQL query into a set of *functional tuple generating dependencies (f-tgd)*, an extension of the classic tgd in which functional expressions or deterministic computations are permitted in the head to generate new tuple values.

```python
import re

def sql_to_ftgd(sql: str, schema: dict):
    sql = " ".join(sql.strip().split())

    def validate_attrs(table, attrs):
        if table not in schema:
            return attrs

        table_cols = schema[table]
        for a in attrs:
            if a not in table_cols:
                raise ValueError(f"Attribute {a} not in table {table}")
        return attrs

    where_pattern = re.compile(
        r"SELECT\s+(.*?)\s+FROM\s+(\w+)\s+WHERE\s*\((.*)\)",
        re.IGNORECASE | re.DOTALL
    )

    m = where_pattern.match(sql)
    if m:
        select_clause, table, where_clause = m.groups()

        parts = [p.strip() for p in select_clause.split(",")]
        body = f"{table}({', '.join(parts)})"

        conditions = [c.strip() for c in where_clause.split("AND")]

        head_conditions = ", ".join(conditions)
        head = f"{head_conditions} -> result({', '.join(parts)})"

        return [f"{body}, {head}"]

    # --------------------------------------------------
    # CASE 1: Aggregation using UNION
    # --------------------------------------------------
    union_agg_pattern = re.compile(
        r"""
        SELECT\s+(\w+),\s*(\w+)\((\w+)\)\s+AS\s+(\w+)
        \s+FROM\s*\(
            SELECT\s+(\w+),\s*(\w+)\s+FROM\s+(\w+)
            \s+UNION\s+
            SELECT\s+(\w+),\s*(\w+)\s+FROM\s+(\w+)
        \)\s+\w+
        \s+GROUP\s+BY\s+(\w+)
        """,
        re.IGNORECASE | re.VERBOSE
    )

    m = union_agg_pattern.match(sql)
    if m:
        group_attr, alias, agg_func, agg_attr, \
        t1_attr1, t1_attr2, table1, \
        t2_attr1, t2_attr2, table2, \
        group_by = m.groups()

        if group_attr != group_by:
            raise ValueError("GROUP BY mismatch")

        validate_attrs(table1, [t1_attr1, t1_attr2])
        validate_attrs(table2, [t2_attr1, t2_attr2])

        body1 = f"{table1}({t1_attr1}, {t1_attr2})"
        body2 = f"{table2}({t2_attr1}, {t2_attr2})"

        head = f"{alias} = {agg_func}({agg_attr}) -> result(pid, {group_attr}, {alias})"

        return [
            f"{body1}, {head}",
            f"{body2}, {head}"
        ]

    # --------------------------------------------------
    # CASE 2: Aggregation (GROUP BY multiple attributes)
    # --------------------------------------------------
    agg_pattern = re.compile(
        r"""
        SELECT\s+(.*?)\s*,\s*
        (\w+)\((\w+)\)\s+AS\s+(\w+)
        \s+FROM\s+(\w+)\s+
        GROUP\s+BY\s+(.*)
        """,
        re.IGNORECASE | re.VERBOSE
    )

    m = agg_pattern.match(sql)
    if m:
        select_attrs, agg_func, agg_attr, alias, table, group_by = m.groups()

        select_attrs = [a.strip() for a in select_attrs.split(",")]
        group_by_attrs = [a.strip() for a in group_by.split(",")]

        validate_attrs(table, select_attrs + group_by_attrs + [agg_attr])

        body_attrs = list(dict.fromkeys(group_by_attrs + [agg_attr]))
        body = f"{table}({', '.join(body_attrs)})"

        head = f"{alias} = {agg_func}({agg_attr}) -> result({', '.join(group_by_attrs + [alias])})"

        return [f"{body}, {head}"]

    # --------------------------------------------------
    # CASE 3: UNION ALL (with the complete schema in the body)
    # --------------------------------------------------
    union_all_pattern = re.compile(
        r"SELECT\s+(.*?)\s+FROM\s+(\w+)",
        re.IGNORECASE
    )

    selects = union_all_pattern.findall(sql)

    if len(selects) >= 2:
        results = []

        for select_clause, table in selects:
            parts = [p.strip() for p in select_clause.split(",")]

            time_attr = None
            type_attr = None
            region_value = None
            intensity_attr = None

            for p in parts:
                m_const = re.match(r"'([^']+)'\s+AS\s+(\w+)", p, re.IGNORECASE)
                if m_const:
                    value, alias = m_const.groups()
                    if alias.lower() == "region":
                        region_value = value
                    continue

                m_alias = re.match(r"(\w+)\s+AS\s+(\w+)", p, re.IGNORECASE)
                if m_alias:
                    attr, alias = m_alias.groups()
                    if alias.lower() == "intensitymean":
                        intensity_attr = attr.lower()
                    continue

                if p.lower() == "time":
                    time_attr = "time"
                elif p.lower() == "type":
                    type_attr = "type"

            if table in schema:
                full_attrs = ", ".join(schema[table])
            else:
                full_attrs = ""

            body = f"{table}({full_attrs})"
            head = f"IntensityMean = {intensity_attr} -> result({time_attr}, {type_attr}, '{region_value}', IntensityMean)"
            results.append(f"{body}, {head}")

        return results

    # --------------------------------------------------
    # CASE 4: Projection using functions (e.g., ROUND, CAST)
    # --------------------------------------------------
    round_pattern = re.compile(
        r"""
        SELECT\s+
        (?P<pre_attrs>\w+(?:\s*,\s*\w+)*)
        \s*,\s*
        (?P<func>\w+)\s*\(\s*
            (?P<attr2>\w+)(?:::\w+)?\s*,\s*
            (?P<precision>\d+)\s*\)
        \s+AS\s+(?P<alias>\w+)\s+
        FROM\s+(?P<table>\w+)
        """,
        re.IGNORECASE | re.VERBOSE
    )

    m = round_pattern.match(sql)
    if m:
        pre_attrs = [a.strip() for a in m.group('pre_attrs').split(',')]
        func      = m.group('func')
        attr2     = m.group('attr2')
        alias     = m.group('alias')
        table     = m.group('table')

        validate_attrs(table, pre_attrs + [attr2])

        body = f"{table}({', '.join(pre_attrs + [attr2])})"
        head = f"{alias} = {func}({attr2}) -> result({', '.join(pre_attrs + [alias])})"

        return [f"{body}, {head}"]

    raise ValueError("Unsupported SQL format")
```

```python
def load_schema(cursor):
    cursor.execute("""
        SELECT table_name, column_name
        FROM information_schema.columns
        ORDER BY table_name, ordinal_position
    """)

    schema = {}
    for table, column in cursor.fetchall():
        schema.setdefault(table, []).append(column)

    return schema
```

### Combine two f-tgds
In the following two sets of f-tgds are combined into a single set of f-tgds.

```python
import re

def extract_head(ftgd: str):
    left, _ = ftgd.split("->", 1)
    m = re.search(r"(\w+\([^)]*\))", left)
    return m.group(1), left.strip()

def extract_right_head(ftgd: str):
    _, right = ftgd.split("->", 1)
    return right.strip()

def combine_ftgds(q1: list[str], q2: list[str]) -> list[str]:

    q1_head, _ = extract_head(q1[0])

    combined = []

    for f in q2:
        left, _ = f.split("->", 1)
        left = left.strip()

        right_head = extract_right_head(f)

        if right_head.startswith("result"):
            combined.append(f"{left} -> {q1_head}")
        else:
            combined.append(f)

    combined.append(q1[0])

    return combined

query_parsed = combine_ftgds(query_parsed_2, query_parsed_1)
query_parsed = combine_ftgds(query_parsed_3, query_parsed)
query_parsed = combine_ftgds(query_parsed_4, query_parsed)

print("=== QUERY (set of t-tgds) ===")
for r in query_parsed:
    print(r)
```

### f-tgd -> SQL
In the following, a set of f-tgds is parsed into one nested SQL query.

```python
import re

def parse_ftgds_to_sql(ftgds: str, table_name="ti"):
    queries = []

    lines = [line.strip() for line in ftgds.strip().split("\n") if line.strip()]

    for line in lines:
        match = re.search(r'IntensityMean\s*=\s*(r\d+_\d+)', line, re.IGNORECASE)
        if not match:
            continue

        source_attr = match.group(1)
        region_label = source_attr.upper()

        query = f"""SELECT time, type, '{region_label}' AS region, {source_attr} AS intensityMean
FROM {table_name}"""

        queries.append(query)

    return "\nUNION ALL\n".join(queries)
```

```python
import re
from typing import List, Tuple, Union, Set

def ftgd_to_sql(ftgd_input: Union[str, List[str]]) -> Tuple[str, List[str]]:

    ftgd_list = [ftgd_input] if isinstance(ftgd_input, str) else ftgd_input

    mappings:      dict[str, List[Tuple[str, List[str]]]] = {}
    mapping_filt:  dict[str, List[str]] = {}
    aggregations:  List[Tuple[str, List[str], str, str, str, str, List[str]]] = []
    group_bys:     List[str] = []

    def split_attrs(s: str) -> List[str]:
        return [a.strip() for a in s.split(",") if a.strip()]

    def infer_region(attr: str) -> str:
        m = re.match(r"r(\d+)_(\d+)", attr)
        return f"R{m.group(1)}_{m.group(2)}" if m else ""

    map_pat    = re.compile(r"^(\w+)\(([^)]*)\)\s*->\s*(\w+)\(([^)]*)\)$")
    assign_pat = re.compile(r"^(\w+)\(([^)]*)\),\s*(\w+)\s*=\s*(\w+)\s*->\s*(\w+)\(([^)]*)\)$")
    agg_pat    = re.compile(r"^(\w+)\(([^)]*)\)\s*,\s*(\w+)\s*=\s*(\w+)\((\w+)\)\s*->\s*(\w+)\(([^)]*)\)$")
    filter_pat = re.compile(r"^(\w+)\(([^)]*)\)\s*,\s*(.+?)\s*->\s*(\w+)\(([^)]*)\)$")

    for ftgd in ftgd_list:
        ftgd = " ".join(ftgd.strip().split())

        m = map_pat.match(ftgd)
        if m:
            src, src_attrs, tgt, tgt_attrs = m.groups()
            mappings.setdefault(tgt, []).append((src, split_attrs(src_attrs)))
            continue

        m = assign_pat.match(ftgd)
        if m:
            src, src_attrs, new_attr, src_attr, tgt, tgt_attrs = m.groups()
            src_attrs = split_attrs(src_attrs)
            tgt_attrs = split_attrs(tgt_attrs)

            mapping_attrs = []
            for ta in tgt_attrs:
                if ta.lower() == new_attr.lower():
                    mapping_attrs.append(src_attr)
                else:
                    for sa in src_attrs:
                        if sa.lower() == ta.lower():
                            mapping_attrs.append(sa)
                            break

            mappings.setdefault(tgt, []).append((src, mapping_attrs))
            continue

        m = agg_pat.match(ftgd)
        if m:
            src, src_attrs, new_attr, func, func_attr, tgt, tgt_attrs = m.groups()
            aggregations.append((
                src,
                split_attrs(src_attrs),
                new_attr,
                func,
                func_attr,
                tgt,
                split_attrs(tgt_attrs)
            ))
            continue

        m = filter_pat.match(ftgd)
        if m:
            src, src_attrs, condition, tgt, tgt_attrs = m.groups()
            for cond in split_attrs(condition):
                mapping_filt.setdefault(tgt, []).append(cond)

            mappings.setdefault(tgt, []).append((src, split_attrs(src_attrs)))
            continue

        raise ValueError(f"Unbekannter FTGD-Ausdruck: {ftgd}")

    def build_mapping_chain() -> Tuple[str, str]:
        tgt_order: List[str] = []
        seen: Set[str] = set()

        for ftgd in ftgd_list:
            ftgd = " ".join(ftgd.strip().split())

            if agg_pat.match(ftgd):
                continue

            for pat in (map_pat, assign_pat, filter_pat):
                m = pat.match(ftgd)
                if m:
                    tgt = m.groups()[-2]
                    if tgt not in seen:
                        seen.add(tgt)
                        tgt_order.append(tgt)

        current_sql: str | None = None
        current_tgt: str | None = None
        current_attrs: List[str] | None = None

        for tgt in tgt_order:
            entries = mappings[tgt]
            src = entries[0][0]
            conds = mapping_filt.get(tgt, [])

            if src.lower() == "ti" and len(entries) > 1:
                union_parts = []

                for _, attrs in entries:
                    parts = []
                    region = None
                    intensity = None

                    for a in attrs:
                        if re.match(r"r\d+_\d+", a):
                            region = infer_region(a)
                            intensity = a
                        else:
                            parts.append(a)

                    if region and intensity:
                        parts.append(f"'{region}' AS region")
                        parts.append(f"{intensity} AS intensityMean")

                    union_parts.append(
                        f"SELECT {', '.join(parts)}\nFROM {src}"
                    )

                current_sql = "\nUNION ALL\n".join(union_parts)

            else:
                attrs: List[str] = []
                for _, a in entries:
                    attrs.extend(a)

                if current_sql is None:
                    current_sql = f"SELECT {', '.join(attrs)}\nFROM {src}"
                    current_attrs = attrs
                else:
                    if current_attrs is not None and set(attrs) == set(current_attrs):
                        pass
                    else:
                        current_sql = (
                            f"SELECT {', '.join(attrs)}\nFROM (\n{current_sql}\n) AS {src}"
                        )
                        current_attrs = attrs

            if conds:
                current_sql = (
                    f"SELECT {', '.join(current_attrs)}\n"
                    f"FROM (\n{current_sql}\n) AS {tgt}\n"
                    f"WHERE ({' AND '.join(conds)})"
                )

            current_tgt = tgt

        return current_sql, current_tgt

    current_sql, current_tgt = build_mapping_chain()

    for src, src_attrs, new_attr, func, func_attr, tgt, tgt_attrs in aggregations:
        sel_parts: List[str] = []
        group_cols: List[str] = []

        for attr in tgt_attrs:
            if attr == new_attr:
                sel_parts.append(f"{func.upper()}({func_attr}) AS {new_attr}")
            else:
                sel_parts.append(attr)
                group_cols.append(attr)

        current_sql = (
            f"SELECT {', '.join(sel_parts)}\n"
            f"FROM (\n{current_sql}\n) AS {src}"
        )

        if func.upper() != "ROUND" and group_cols:
            current_sql += f"\nGROUP BY {', '.join(group_cols)}"
            group_bys = group_cols

        current_tgt = tgt

    return current_sql.strip(), group_bys
```

### Provenance
In the following, provenance information are calculated. Each tuple in the query result is assigend a *witness base* (why) and a *provenance polynomial* (how). In addition, a *list of all provenance IDs* involved in the result (where) is provided.

```python
import re
from typing import List, Tuple

def extract_select_clauses(sql: str) -> List[Tuple[int, int, str]]:
    pattern = re.compile(r'\bSELECT\b', re.IGNORECASE)
    matches = list(pattern.finditer(sql))
    if not matches:
        return []
    clauses = []
    for i, m in enumerate(matches):
        start = m.start()
        end = matches[i + 1].start() if i + 1 < len(matches) else len(sql)
        clauses.append((start, end, sql[start:end]))
    return clauses

def add_provenance_to_select(clause: str, prov_id: str) -> str:
    low = clause.lower()
    if prov_id.lower() in low or "src" in low:
        return clause
    sel_pat = re.compile(r'(?i)(SELECT\s+(?:DISTINCT|ALL)?\s*)')
    return sel_pat.sub(r'\1' + prov_id + ', ', clause, count=1)

def add_provenance_to_group_by(sql: str, prov_id: str) -> str:
    pat = re.compile(r'(?i)GROUP\s+BY\s+([^\n;]+)', re.DOTALL)
    def repl(m):
        attrs_text = m.group(1).strip()
        attrs = [a.strip() for a in attrs_text.split(',') if a.strip()]
        low_attrs = [a.lower() for a in attrs]
        if prov_id.lower() not in low_attrs:
            attrs.append(prov_id)
        return f"GROUP BY {', '.join(attrs)}"
    return pat.sub(repl, sql)

def add_provenance_to_all(sql: str, prov_id: str) -> str:
    sel_clauses = extract_select_clauses(sql)
    if not sel_clauses:
        return sql
    modified_selects = [add_provenance_to_select(t, prov_id) for (_, _, t) in sel_clauses]

    parts, prev_end = [], 0
    for (start, end, _), new_clause in zip(sel_clauses, modified_selects):
        parts.append(sql[prev_end:start])
        parts.append(new_clause)
        prev_end = end
    parts.append(sql[prev_end:])
    updated_sql = "".join(parts)

    updated_sql = add_provenance_to_group_by(updated_sql, prov_id)
    return updated_sql

def extract_subqueries(sql: str) -> List[str]:
    subqueries, stack = [], []
    for i, ch in enumerate(sql):
        if ch == '(':
            stack.append(i)
        elif ch == ')':
            if stack:
                start = stack.pop()
                inner = sql[start+1:i].strip()
                if re.search(r'\bSELECT\b', inner, re.IGNORECASE):
                    subqueries.append(inner)
    return subqueries
```

```python
import re
from typing import List, Tuple, Optional
from collections import OrderedDict

def fetch_provenance(
        cursor,
        prov_query: str,
        provenance_col: str = "provid",
        type_col: str = "type"
):
    cursor.execute(prov_query)
    rows = cursor.fetchall()

    col_names = [desc[0] for desc in cursor.description]

    try:
        idx_pid = col_names.index(provenance_col)
        idx_type = col_names.index(type_col)
    except ValueError as exc:
        raise ValueError(
            f"Column '{provenance_col}' or '{type_col}' not found."
        ) from exc

    return [
        (row[idx_pid], row[idx_type])
        for row in rows
        if row[idx_pid] is not None
    ]

def compute_provenance_from_result(
        cursor,
        query_text: str,
        from_clause: str,
        prov_result,
        copy_times: int = 1
):
    from collections import OrderedDict

    pid_type_list = [item for group in prov_result for item in group]
    unique_pid_types = list(OrderedDict.fromkeys(pid_type_list))

    where = []
    why = {}
    how = {}

    for pid, typ in unique_pid_types:

        # WHERE: now per row, not grouped
        where.append((typ, pid))

        # WHY: now PID-based (IMPORTANT FIX)
        why.setdefault(typ, []).append([pid for _ in range(copy_times)])

        # HOW: PID-based
        expr = " + ".join([pid] * copy_times)
        how.setdefault(typ, []).append(expr)

    return where, why, how

def count_region_columns(sql: str) -> int:
    return len(re.findall(r'\bAS\s+region\b', sql, flags=re.IGNORECASE))

def process_all_subqueries(
        cursor,
        full_sql: str,
        provenance_col: str = "provid",
        copy_times: Optional[int] = None
) -> Tuple[List[str], List[List[List[str]]], List[List[str]]]:
    if copy_times is None:
        copy_times = count_region_columns(full_sql) or 1
    last_where_where: Optional[List[str]] = None
    last_where_why:   Optional[List[List[List[str]]]] = None
    last_where_how:   Optional[List[List[str]]] = None
    where: List[str] = []
    why:   List[List[List[str]]] = []
    how:   List[List[str]] = []
    for idx, subq in enumerate(subqueries, 1):
        try:
            prov_ids = fetch_provenance(cursor, subq, provenance_col)
        except Exception as exc:
            print(f"❌ Error in the provenance query: {exc}")
            continue
        from_clause = f"FROM ({subq}) AS sub{idx}"
        where, why, how = compute_provenance_from_result(
            cursor=None,
            query_text=subq,
            from_clause=from_clause,
            prov_result=[prov_ids],
            copy_times=copy_times
        )
        def has_top_level_where(sql: str) -> bool:
            depth = 0
            tokens = re.split(r'(\(|\))', sql)
            for token in tokens:
                if token == '(':
                    depth += 1
                elif token == ')':
                    depth -= 1
                else:
                    if depth == 0 and re.search(r'\bWHERE\b', token, re.IGNORECASE):
                        return True
            return False
        if has_top_level_where(subq):
            last_where_where = where
            last_where_why   = why
            last_where_how   = how
        else:
            if last_where_where is not None:
                where = last_where_where
                why   = last_where_why
                how   = last_where_how
    return where, why, how
```

### XML
Finally, all relevant information is written to an XML file. This includes:
- database schema (relevant for answering the query)
- original database instance (relevant for answering the query)
- query result
- provenance result

```python
import re

def extract_table_names(from_clause):
    matches = re.findall(
        r"\bFROM\s+([A-Za-z_][A-Za-z0-9_]*)|\bJOIN\s+([A-Za-z_][A-Za-z0-9_]*)",
        from_clause,
        re.IGNORECASE
    )

    tables = set()
    for a, b in matches:
        if a:
            tables.add(a)
        if b:
            tables.add(b)

    return tables


def get_relation_schemas(cursor, from_clause):
    full_schema = load_schema(cursor)
    used_tables = extract_table_names(from_clause)

    relation_schemas = {
        table: full_schema[table]
        for table in used_tables
        if table in full_schema
    }

    return relation_schemas

relation_schemas = get_relation_schemas(cursor, query)
```

```python
import xml.etree.ElementTree as ET

def write_provenance_xml(
        filename,
        relation_schemas,
        base_data,
        result,
        where,
        why,
        how
):
    root = ET.Element("input")

    # -----------------------
    # SCHEMA
    # -----------------------
    schema_el = ET.SubElement(root, "schema")
    relations_el = ET.SubElement(schema_el, "relations")

    for rel, cols in relation_schemas.items():
        rel_el = ET.SubElement(relations_el, "relation", {"name": rel})
        for c in cols:
            ET.SubElement(rel_el, "attribute", {"name": c})

    if result:
        res_cols = result[0].keys()
        res_rel = ET.SubElement(relations_el, "relation", {"name": "result"})
        for c in res_cols:
            ET.SubElement(res_rel, "attribute", {"name": c})

    # -----------------------
    # ORIGINAL INSTANCE
    # -----------------------
    original_el = ET.SubElement(root, "originalinstance")

    # where is now: [(typ, pid), ...]
    where_ids = {pid for _, pid in where}
    written_ids = set()

    for rel, rows in base_data.items():
        for row in rows:
            provid = row.get("provid")
            if provid is None:
                continue

            if provid not in where_ids or provid in written_ids:
                continue

            written_ids.add(provid)

            atom = ET.SubElement(original_el, "atom", {
                "name": rel,
                "id": provid
            })

            for k, v in row.items():
                ET.SubElement(atom, "constant", {
                    "name": k,
                    "value": str(v)
                })

    # -----------------------
    # RESULT INSTANCE
    # -----------------------
    instance_el = ET.SubElement(root, "instance")

    for i, row in enumerate(result, start=1):
        atom = ET.SubElement(instance_el, "atom", {
            "name": "result",
            "id": f"result_{i}"
        })

        for k, v in row.items():
            ET.SubElement(atom, "constant", {
                "name": k,
                "value": str(v)
            })

    # -----------------------
    # PROVENANCE
    # -----------------------
    prov_el = ET.SubElement(root, "provenance")

    # pid → type mapping (correct for list-based where)
    pid_to_type = {pid: typ for typ, pid in where}

    # iterate directly over list (NO .items(), NO flattening)
    for i, (typ, pid) in enumerate(where):

        atom = ET.SubElement(prov_el, "atom", {
            "name": "result",
            "id": f"result_{i+1}"
        })

        # -----------------------
        # WHERE
        # -----------------------
        where_el = ET.SubElement(atom, "where")
        tuples_el = ET.SubElement(where_el, "tuples")
        ET.SubElement(tuples_el, "tuple", {"id": pid})

        # -----------------------
        # TYPE
        # -----------------------
        ET.SubElement(atom, "constant", {
            "name": "source_type",
            "value": typ
        })

        # -----------------------
        # WHY
        # -----------------------
        why_el = ET.SubElement(atom, "why")
        wb_el = ET.SubElement(why_el, "witness_base")

        for witness in why.get(typ, []):

            w_el = ET.SubElement(wb_el, "witness")

            for tid in witness:
                if isinstance(tid, list):
                    for x in tid:
                        ET.SubElement(w_el, "tuple", {"id": str(x)})
                else:
                    ET.SubElement(w_el, "tuple", {"id": str(tid)})

        # -----------------------
        # HOW
        # -----------------------
        how_el = ET.SubElement(atom, "how")
        expr = how.get(typ, [""])[0] if how.get(typ) else ""
        ET.SubElement(how_el, "polynom", {"expr": expr})

    # -----------------------
    # WRITE FILE
    # -----------------------
    tree = ET.ElementTree(root)
    tree.write(filename, encoding="utf-8", xml_declaration=True)

    print(f"✅ XML saved at: {filename}")
```
